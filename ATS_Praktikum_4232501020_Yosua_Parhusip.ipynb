```python
#Nama: Yosua Parhusip
#NIM: 4232501020
#Tanggal: 09 November 2026
#Mata Kuliah: Pemrograman Dasar

#Soal No 10
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for i, suhu in enumerate(data, start=1):
    try:
        nilai = float(suhu)
        jumlah_valid += 1

        if nilai > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

print(f"Jumlah data valid: {jumlah_valid}")
```
