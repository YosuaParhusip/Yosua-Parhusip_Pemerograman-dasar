```python
#Soal Esai

1. Dalam bahasa pemrograman python, ekspresi 10 + 5* 2 menghasilkan nilai 20, bukan 30. Hal ini terjadi karena python
menerapkan aturan prioritas operator (operator precedence) yang menentukan operator mana yang akan dieksekusi terlebih dahulu
dalam sebuah ekspresi.

Python nggak proses ekspresi secara linear kiri-ke-kanan doang. Dia jalanin berdasarkan hierarki prioritas operator,
mirip prinsip precedence di compiler, bukan cuma "baca kalimat". Di tabel precedence, * levelnya lebih tinggi dari +. Artinya * "bind"
duluan ke operand-nya sebelum + sempat dieksekusi, walau posisinya di tengah/kanan.Jadi 10 + 5 * 2 itu secara implisit di-parse jadi:
10 + (5 * 2) bukan (10 + 5) * 2.
Alurnya:
5 * 2 = 10 → dieksekusi duluan (prioritas tinggi)
10 + 10 = 20 → baru operator prioritas rendah jalan
Kalau logikanya cuma left-to-right tanpa precedence (kayak asumsi orang awam), ya keluar 30. Tapi itu salah kaprah,
Python (dan hampir semua bahasa pemrograman) itu evaluasinya berbasis grammar/precedence rule, bukan urutan penulisan.
Sumber : https://belajarpython.com/tutorial/operator-python/
```

```python
2. Perbedaan mendasar antara operator relasional dan operator logika terletak pada tujuan penggunaannya.
Tujuan operator relasional adalah membandingkan dua nilai atau variabel (angka, string, dsb) sedangkan tujuan operator logika
adalah mengkombinasi atau membalikan nilai kebenaran (boolean).

Contoh Situasi: Sebuah bank menggunakan sistem otomatis untuk menentukn apakah seorang nasabah layak mendapatkan pinjaman.

kriterianya adalah:
1. Usia nasabah harus minimal 25 tahun.
2. Pendapatan bulanan harus lebih dari Rp.7.000.000.
3. Skor kredit nasabah tidak boleh negatif (palsu/bermasalah).

Dalam situasi ini,kita menggunakan operator relasional untuk memeriksa setiap kriteria angka, dan operator logika untuk
menggabungkan semua syarat tersebut.

usia = 27
pendapatan =9000000
skor_kredit_buruk = False

# Ekspresi gabungan menggunakan operator relasional dan logika
layak_pinjam = (usia >= 25) and (pendapatan > 7000000) and (not skor_kredit_buruk)
print(layak_pinjam)
```

```python
3. Kesalahan operator precedence terjadi saat ingin menghitung bagian gaji yang diterima oleh dua orang karyawan setelah
dipotong pajak bulanan. Jika tidak menggunakan tanda kurung (), operasi pembagian  akan dieksekusi terlebih dahulu sebelum pengurangan,
sehingga menghasilkan nilai yang salah.
```

```python
# Bukti Jawaban 3

# Bukti Jawaban 3
gaji_kotor = 10000000
pajak = 1000000

# Salah: Pembagian (/) dieksekusi lebih dulu daripada pengurangan (-)
total_per_orang = gaji_kotor - pajak / 2
print(total_per_orang)  # Hasilnya: 9500000.0 (Salah, karena hanya pajak yang dibagi 2)


gaji_kotor = 10000000
pajak = 1000000

# Benar: Tanda kurung memaksa pengurangan dihitung terlebih dahulu
total_per_orang = (gaji_kotor - pajak) / 2
print(total_per_orang)  # Hasilnya: 4500000.0 (Benar)
```

```python
# Nama    : Yosua Parhusip
# NIM     : 4232501020
# Tanggal : 25 September 2026
# Tujuan  : Melakukan validasi logis terhadap kelayakan seseorang untuk mengemudi sendiri berdasarkan hukum dan kondisi keselamatan.

# Jawaban No 3

# Meminta input umur dari pengguna
umur = int(input("Masukkan umur anda: "))

# Meminta input status SIM (True/False)
input_sim = input("Apakah anda punya SIM? (True/False): ")
punya_sim = input_sim.strip().lower() == 'true'

# Meminta input tambahan untuk contoh operator precedence (izin khusus/darurat)
input_KTP = input("Apakah punya KTP? (True/False); ")
KTP = input_KTP.strip().lower() == 'true'

# Ekspresi dengan tanda kurung (operator precedence)
# Tanpa tanda kurung: umur >= 17 or KTP dan punya_sim
# (karena 'and' lebih kuat dari 'or', program akan mengira: umur >= 17 or (KTP and punya_sim))
# Dengan tanda kurung di bawah, kita memaksa agar (umur >= 17 atau KTP) dievaluasi terlebih dahulu.
boleh_mengemudi = (umur >= 17 or KTP) and punya_sim

#Menampilkan hasil keputusan
if boleh_mengemudi:
    print("Hasil: Anda Boleh mengemudi sendiri.")
else:
    print("Hasil: Anda TIDAK boleh mengemudi sendiri")
```
