```python
# Nama: Yosua Parhusip
# Nim : 4232501020

#soal 1 (belanja buku)

jumlah_buku = 7
harga_per_buku = 18500
diskon = 10 / 100
uang_budi = 150000

#total belanja
total_belanja = jumlah_buku * harga_per_buku

#diskon
total_diskon = total_belanja * diskon

#yang harus dibayar
jumlah_dibayar = total_belanja - total_diskon

#uang kembalian
uang_kembalian = uang_budi - jumlah_dibayar

#iuran per orang
jumlah_orang = 4
iuran_per_orang = jumlah_dibayar // jumlah_orang
sisa = jumlah_dibayar % jumlah_orang
uang_cukup = uang_budi >= jumlah_dibayar

#tampilkan hasil perhitungann
print(f"total belanja :", total_belanja)
print(f"total diskon :", total_diskon)
print(f"jumlah yang dibayar :", jumlah_dibayar)
print(f"uang kembalian :", uang_kembalian)
print(f"iuran per orang :", iuran_per_orang)
print(f"sisa :", sisa)
print(f"uang cukup :", uang_cukup)
```

```python
#soal 2 (PLTS atap rumah)

jumlah_panel = 12
daya = 450
penyinaran = 5
efesiensi = 0.8
kebutuhan_listrik_rumah = 25
tarif_listrik = 1444.70

#energi yang di hasilkan per hari (KWh)
total_daya = jumlah_panel * daya
energi_harian = (total_daya * penyinaran * efesiensi) / 1000

#energi per bulan
energi_bulan = energi_harian * 30
nilai_listrik = energi_bulan * tarif_listrik

#kekurangan energi per hari terhadap kebutuhan
kekurangan_energi = kebutuhan_listrik_rumah - energi_harian

#apakah PLTS sudah mencukupi kebutuhan rumah
plts_mencukupi = energi_harian >= kebutuhan_listrik_rumah

#tampilkan hasil
print(f"total daya :", total_daya)
print(f"energi harian :", energi_harian)
print(f"energi bulan :", energi_bulan)
print(f"nilai listrik :", nilai_listrik)
print(f"kekurangan energi :", kekurangan_energi)
print(f"plts mencukupi :", plts_mencukupi)
```

```python
#soal 3 (generator diesel dan rugi kabel)

#generator diesel
daya_generator = 200  # kW
jam_operasi = 8       # jam/hari
konsumsi_solar = 0.25 # liter/kWh
harga_solar = 13500   # Rp/liter
kapasitas_tangki = 1000 # liter
arus = 20             # A
hambatan = 0.5        # Ohm

#1. energi yang dibangkitkan per hari (kWh)
energi = daya_generator * jam_operasi

#2. kebutuhan solar per hari dan biayanya
solar_per_hari = energi * konsumsi_solar
biaya_per_hari = solar_per_hari * harga_solar

#3. jumlah hari penuh tangki cukup dan sisa solar (menggunakan // dan %)
hari_cukup = kapasitas_tangki // solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari

#4. rugi daya pada kabel (P = I^2 * R) menggunakan operator **
rugi_kabel = (arus ** 2) * hambatan

#5. evaluasi apakah rugi kabel tinggi (> 150 W dan arus >= 20 A)
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

#menampilkan output sesuai format contoh
print(f"Energi (kWh)   : {energi}")
print(f"Solar (liter)  : {solar_per_hari}")
print(f"Biaya (Rp)     : {biaya_per_hari}")
print(f"Tangki cukup   : {float(hari_cukup)} hari")
print(f"Sisa solar (L) : {sisa_solar}")
print(f"Rugi kabel (W) : {rugi_kabel}")
print(f"Rugi tinggi?   : {rugi_tinggi}")
```
