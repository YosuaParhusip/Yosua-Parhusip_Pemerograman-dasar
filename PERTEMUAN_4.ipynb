```python
#latihan 3
harga_gorengan = 2000
jumlah_gorengan = 3
harga_esteh = 4000
jumlah_esteh = 2
uang = 20000

#total belanja
total = harga_gorengan * jumlah_gorengan + harga_esteh * jumlah_esteh

#uang_kembalian
kembalian = uang - total

#status uang cukup
cukup = uang >= total
```

```python
#tampilkan_hasil_perhitungan

print("total_belanja  :", total, "rupiah")
print("uang_kembalian :", kembalian, "rupiah")
print("uang_cukup?    :", cukup)
```

dina menonton film maraton dengan total durasi 275 menit.

tentukan:
1. berapa jam penuh dan berapa menit sisa (dalam format ".... jam .... menit").
2. total durasi dalam jam desimal.
3. apakah durasi tersebut lebih dari 3 jam dan kurang dari 5 jam? (True/False)

jam penuh = total menit dibagi 60 tanpa pecahan (//).
menit sisa = sisa pembagian total menit dengan 60 (%).
jam desimal = total menit dibagi 60 dengan pecahan (/).
cek rentang dengan > dan < yang digabung and.

```python
#latihan 4

total_menit = 275

#operasikan
jam = total_menit // 60
sisa_menit = total_menit % 60
jam_desimal = total_menit / 60
dalam_rentang = total_menit >= 180 and total_menit <= 300

#tampilkan_hasil_operasi
print("durasi :", jam, "jam", sisa_menit, "menit")
print("jam desimal :", round(jam_desimal, 2))
print("3 s.d. 5 jam?:", dalam_rentang)
```
